```python
import json

prefix = "1930_match_1_"
matches = []

with open("match_chunks.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        item = json.loads(line)

        if item["id"].startswith(prefix):
            matches.append(item)

print("Number of matches:", len(matches))

for item in matches:
    print(
        item["id"],
        "=>",
        item["metadata"].get("chunk_type")
    )
```
